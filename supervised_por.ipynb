```python
import numpy as np
import pandas as pd
import ipywidgets as widgets
from ipywidgets import Dropdown, Output, HBox, VBox
from IPython.display import display, clear_output
from sklearn.datasets import make_classification, make_regression
from sklearn.model_selection import train_test_split
np.random.seed(42)
```

```python
#loading math and portugues csv files as datasets to be used as multiple linear regression 
df1 = pd.read_csv("./student-mat.csv", sep=";")
df2 = pd.read_csv("./student-por.csv", sep=";")
```

MATH

```python
features_mat = ["G1", "G2"]

X = df1[features_mat]
Y = df1["G3"]
```

```python
#training and testing our variables
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
#determing the r2 score to see how well the X values predict the y values 

from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print(r2)
```
